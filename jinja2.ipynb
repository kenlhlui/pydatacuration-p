```python
template_dict = {
    "file_open": {
        "status": {},
        "comments": [],
        "not_checked": []
    },
    "special_characters": {
        "status": {},
        "comments": []
    },
    "long_file_length": {
        "status": {},
        "comments": []
    },
    "file_ext": {
        "status": {},
        "comments": []
    },
    "readme_file": {
        "status": {},
        "comments": []
    },
    "typo": {
        "status": {},
        "comments": []
    },
    "missing_field": {
        "title": {
            "status": {},
            "comments": []
        },
        "description": {
            "status": {},
            "comments": []
        },
        "subject": {
            "status": {},
            "comments": []
        },        
        "author_name": {
            "status": {"SI": "X"},
            "comments": []
        },
        "author_affiliation": {
            "status": {},
            "comments": []
        }
    }
}
print(template_dict)
```

```python
from jinja2 import Environment, FileSystemLoader
```

```python
def read_csv_template(file):
    with open(file, 'r', encoding='ISO-8859-1') as f:
        content = f.read()
    return content

file = './res/template.csv'
template_string = read_csv_template(file)
```

```python
# load the read_csv_template into a Jinja2 environment

from jinja2 import Environment, FileSystemLoader, Template

env = Environment(loader=FileSystemLoader('/home/ccke9973/git/pydatacuration-p/res/'))

template = Template(template_string)

rendered = template.render(template_dict=template_dict)
```

```python
# Export the rendered template to a file
with open('./workdir/log_files/render_log.csv', 'w') as f:
    f.write(rendered)
```

```python
# Trun the rendered template into a excel file

import pandas as pd
pd.read_csv('./workdir/log_files/render_log.csv').to_excel('./workdir/log_files/render_log.xlsx', index=False)
```
